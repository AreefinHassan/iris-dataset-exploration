(1) Reading the output of datasets.load_iris(). What are the meaning of four attributes/columns? What are class 0, 1, 2 means corresponding to?

Each instance has four features (attributes) and one class label (the species).

Here are the attributes of the Iris dataset:

**Sepal Length:** The length of the sepal in centimeters.

**Sepal Width:** The width of the sepal in centimeters.

**Petal Length:** The length of the petal in centimeters.

**Petal Width:** The width of the petal in centimeters.

And the class label:

**Species:** The species of the iris plant. This is the target variable that we try to predict. The dataset includes three classes:

Iris setosa

Iris virginica

Iris versicolor

```python
# Loading the Iris dataset
X = datasets.load_iris().data
y = datasets.load_iris().target
```

```python
# Printing the dimensions of X
print(X.shape)  # This should print (150, 4), indicating 150 samples and 4 features.
```

```python
y.shape
```

```python
# Reducing dimensions to keep only sepal length and sepal width
new_X = X[:, :2]  # Keeping the first two columns of X

# Checking the shape of new_X
print(new_X.shape)  # This should print (150, 2), as expected.
```

(2) We pick the first two dimensions to reduce dimensions. Create scatter plot to visualize sepal length (1st dimension as X axis) and sepal width (2nd dimension as Y axis), for Setosa and Virginica. (reminder: Python array index starts with 0).

```python
# Scatter plot for Setosa (class 0) and Virginica (class 2)
plt.figure(figsize=(10, 6))
plt.scatter(new_X[y==0, 0], new_X[y==0, 1], s=70, alpha=0.5, label='Setosa')
plt.scatter(new_X[y==2, 0], new_X[y==2, 1], s=70, alpha=0.5, label='Virginica')

plt.xlabel('Sepal Length (cm)')
plt.ylabel('Sepal Width (cm)')
plt.title('Iris Dataset: Sepal Length vs. Sepal Width for Setosa and Virginica')
plt.legend()
plt.show()
```

(3) Do you think they are linear saparable? Why?

Iris Setosa and Iris Virginica appear to be linearly separable based on sepal length and sepal width. The scatter plot shows a clear distinction between the two species with minimal overlap, indicating that a straight line could effectively separate them in this two-dimensional space. This suggests that simple linear classifiers could perform well in distinguishing between these two species using only these two features.

(4) (bonus) Can you use some other dimensions to make it more likely to get linear seperable? Show your scatter plot.

Let's create a scatter plot using petal length (3rd dimension) and petal width (4th dimension) for Setosa and Virginica, and examine their linear separability in this new feature space.

```python
# Using petal length and petal width for the new scatter plot
# Adjusting indices to access the 3rd (petal length) and 4th (petal width) features
new_X_petal = X[:, 2:4]  # Keeping the third and fourth columns of X
```

```python
# Scatter plot for Setosa (class 0) and Virginica (class 2) using petal dimensions
plt.figure(figsize=(10, 6))
plt.scatter(new_X_petal[y==0, 0], new_X_petal[y==0, 1], s=70, alpha=0.5, label='Setosa')
plt.scatter(new_X_petal[y==2, 0], new_X_petal[y==2, 1], s=70, alpha=0.5, label='Virginica')

plt.xlabel('Petal Length (cm)')
plt.ylabel('Petal Width (cm)')
plt.title('Iris Dataset: Petal Length vs. Petal Width for Setosa and Virginica')
plt.legend()
plt.show()
```

The scatter plot using petal length and petal width for Iris Setosa and Iris Virginica reveals an even clearer distinction between the two species compared to using sepal length and width. This visualization shows that:

Setosa is well-separated from Virginica, with distinct clusters for each species.

There is minimal overlap between the two species in this feature space, suggesting a higher degree of linear separability.

This indicates that petal length and petal width are more effective dimensions for linearly separating Setosa and Virginica, supporting the use of these features for simple linear classifiers to achieve high classification accuracy.
